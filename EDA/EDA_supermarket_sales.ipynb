```python
# Task 4 — Feature Engineering
# Create sales_level
df["sales_level"] = pd.cut(
    df["Total"],
    bins=[0, 200, 500, df["Total"].max()],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
# Convert Time
time = pd.to_datetime(df["Time"], format="%H:%M")
df["hour"] = time.dt.hour

# Create day_period
df["day_period"] = pd.cut(
    df["hour"],
    bins=[0, 12, 17, 24],
    labels=["Morning", "Afternoon", "Evening"],
    include_lowest=True,
    right=False
)
print(df[["Total", "sales_level", "Time", "hour", "day_period"]].head())
```

```python
# Numerical EDA

num_cols = [
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Total",
    "cogs",
    "gross margin percentage",
    "gross income",
    "Rating"
]

fig, axes = plt.subplots(len(num_cols), 2, figsize=(14, 35))

for i, col in enumerate(num_cols):

    # Histogram + KDE
    sns.histplot(data=df, x=col, kde=True, ax=axes[i, 0])
    axes[i, 0].set_title(f"Distribution of {col}")

    # Boxplot
    sns.boxplot(data=df, x=col, ax=axes[i, 1])
    axes[i, 1].set_title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

```python
def tukey_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = series[
        (series < lower_bound) |
        (series > upper_bound)
    ]

    return outliers
```

```python
print("Outlier count for each numerical column:\n")

for col in num_cols:
    outliers = tukey_outliers(df[col])

    print(f"{col}: {len(outliers)} outliers")
```

```python
# 7.7 Average Total Sales by Product Line
product_mean = (
    df.groupby("Product line")["Total"]
    .mean()
    .sort_values(ascending=False)
)

print(product_mean)
```
